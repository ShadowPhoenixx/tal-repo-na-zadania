## Zadanie 3: Filtr jako funkcja

**Dane:** Lista liczb całkowitych, która może być pusta i może zawierać powtórzenia.

**Wynik:** Funkcja ma zwrócić nową listę dodatnich liczb, a nie ich sumę ani liczbę wystąpień. Wyświetl zwróconą listę dla `[-2, 0, 5, 5]` oraz pustej listy.

## Zadanie 4: Wiadomość z przesuniętym alfabetem

**Do wykonania:** Napisz funkcję `szyfruj(wiadomosc, k)`. Każda litera ma zostać zastąpiona literą oddaloną o `k` pozycji. Alfabet jest cykliczny: po Z następuje A. Spacje pozostają na swoich miejscach.

**Wynik:** Funkcja zwraca zaszyfrowany napis o takiej samej długości jak wiadomość. Wyświetl wynik dla `"TAJNA BAZA"` i `k = 3`. Wyjaśnij na własnym przykładzie zachowanie na końcu alfabetu.

## Zadanie 6: Wynik i numer wiersza

**Dane:** Lista liczb. Pozycje w wyniku numerujemy od 1, chociaż indeksy Pythona zaczynają się od 0.

**Do wykonania:** Napisz funkcję `pierwsze_max(liczby)`, która znajduje największą wartość i numer jej pierwszego wystąpienia. Lista może być pusta.

**Wynik:** Zwróć parę `(maksimum, numer)`: największą liczbę oraz numer miejsca, na którym pojawia się ona po raz pierwszy. Dla pustej listy zwróć `None`. Zaprezentuj działanie dla `[3, 8, 8, 2]`.

**Przykład wyjaśniający:** Dla `[1, 9, 4, 9]` odpowiedź to `(9, 2)`. Największą wartością jest 9; jej pierwsze wystąpienie jest drugim elementem listy. Późniejsza dziewiątka nie zmienia odpowiedzi. Numer 2 odpowiada indeksowi 1 w Pythonie.


```python
def pierwsze_max(liczby):
    if not liczby:
        return None

    maksimum = max(liczby)
    numer = liczby.index(maksimum) + 1
    return maksimum, numer


print(pierwsze_max([1, 9, 4, 9]))
```

## Zadanie 7: Histogram znaków

**Dane:** Napis, w którym mogą powtarzać się litery i inne znaki. Wielkie i małe litery traktujemy oddzielnie.

**Do wykonania:** Napisz funkcję `histogram(s)` zliczającą wystąpienia każdego znaku. Nie korzystaj z `Counter`.

**Wynik:** Zwróć słownik, w którym każdy napotkany znak ma przypisaną liczbę swoich wystąpień. Pokaż wyniki dla `"o+o*o"`, `"AaA"` i pustego napisu.

**Przykład wyjaśniający:** Dla `"aba!"` wynik to `{"a": 2, "b": 1, "!": 1}`. Znak a występuje dwa razy, a pozostałe znaki po jednym. To zestawienie liczności, nie lista kolejnych znaków. Dla pustego napisu zwróć pusty słownik.


```python
def histogram(s):
    licznosci = {}
    for znak in s:
        licznosci[znak] = licznosci.get(znak, 0) + 1
    return licznosci

print(histogram("o+o*o"))
print(histogram("AaA"))
print(histogram("aba!"))
print(histogram(""))
```

## Zadanie 9: Usuwanie duplikatów z zachowaniem kolejności

**Dane:** Lista liczb, w której mogą wystąpić duplikaty.

**Do wykonania:** Napisz funkcję `unikalne(dane)`, która pozostawia tylko pierwsze wystąpienie każdej wartości. Zachowaj kolejność tych wystąpień. Sam zdecyduj, jakich kolekcji potrzebujesz.

**Wynik:** Zwróć nową listę, w której każda wartość występuje tylko raz. Dla `[4, 2, 4, 7, 2]` kolejność wyniku to `[4, 2, 7]`. Dla pustej listy zwróć `[]`.

**Przykład wyjaśniający:** „Zachowaj kolejność” oznacza kolejność pierwszego pojawienia się wartości, nie kolejność od najmniejszej do największej. Dla `[8, 3, 8, 1, 3]` odpowiedź to `[8, 3, 1]`, a nie `[1, 3, 8]`.


```python
def unikalne(dane):
    wynik = []
    wystapienia = set()

    for liczba in dane:
        if liczba not in wystapienia:
            wynik.append(liczba)
            wystapienia.add(liczba)

    return wynik


print(unikalne([1, 2, 3, 2, 1]))
print(unikalne([4, 2, 4, 7, 2]))
print(unikalne([8, 3, 8, 1, 3]))
```

## Zadanie 10: Dwie podpowiedzi do sejfu

**Dane:** Kod sejfu i próba otwarcia to napisy z czterech różnych cyfr. Zero może znajdować się na początku.

**Do wykonania:** Napisz funkcję `podpowiedz(kod, proba)`. Określ, ile cyfr próby jest poprawnych i stoi na właściwej pozycji, a ile występuje w kodzie, lecz na innej pozycji. Jednej cyfry nie licz w obu grupach.

**Wynik:** Zwróć parę `(na_wlasciwym_miejscu, na_innym_miejscu)` zawierającą dwie liczby cyfr. Wyświetl odpowiedź dla kodu `"5072"` i próby `"5209"`. Uzasadnij obie liczby.

**Przykład wyjaśniający:** Dla kodu `"1234"` i próby `"1428"` wynik to `(1, 2)`: cyfra 1 stoi we właściwym miejscu, cyfry 4 i 2 są w kodzie, lecz w innych miejscach, a cyfry 8 w kodzie nie ma. Nie zwracaj samych cyfr.


```python
def podpowiedz(kod, proba):
    na_wlasciwym_miejscu = 0
    na_innym_miejscu = 0
    for index in range(len(proba)):
        if proba[index] == kod[index]:
            na_wlasciwym_miejscu += 1
        elif proba[index] in kod:
            na_innym_miejscu += 1
    return na_wlasciwym_miejscu, na_innym_miejscu

print(podpowiedz([1, 2, 3, 4], [1, 4, 2, 8]))
print(podpowiedz([5, 0, 7, 2], [5, 2,0, 9]))
```

**Wynik:** Zwróć parę `(lista_palindromow, slownik_licznosci_dlugosci)`. Słownik ma opisywać długości wszystkich napisów wejściowych, nie tylko palindromów. Zaprezentuj raport dla `["aa", "ab", "x", "aa"]`. Możesz wykorzystać własne wcześniejsze funkcje.

**Przykład wyjaśniający:** Dla `["ala", "kot", "xx"]` raport to `(["ala", "xx"], {3: 2, 2: 1})`. Są dwa palindromy, ale napisy o długości 3 liczymy oba, także niebędący palindromem `"kot"`.


```python
def raport_napisow(napisy):
    lista_palindromow = []
    slownik_licznosci_dlugosci = {}
    for napis in napisy:
        if napis == napis[::-1]:
            lista_palindromow.append(napis)
        dlugosc = len(napis)
        if dlugosc in slownik_licznosci_dlugosci:
            slownik_licznosci_dlugosci[dlugosc] += 1
        else:
            slownik_licznosci_dlugosci[dlugosc] = 1
    
    return(lista_palindromow, slownik_licznosci_dlugosci)

print(raport_napisow(["ala", "kot", "xx"]))
```

## Zadanie 12: Kompresja sygnału

**Dane:** Sygnał zapisany jako napis z liter A–Z. Seria to maksymalny spójny fragment jednakowych liter.

**Do wykonania:** Napisz funkcję `spakuj(sygnal)`, która opisuje każdą serię parą: litera i jej liczba wystąpień w tej serii. Oddzielnych serii tej samej litery nie wolno łączyć.

**Wynik:** Zwróć listę par `(litera, dlugosc_serii)` w kolejności sygnału, a dla pustego napisu `[]`. Wyświetl zapis `"AAABBCA"` i wyjaśnij, dlaczego litera A tworzy dwa osobne wpisy.

**Przykład wyjaśniający:** Dla `"CCDC"` wynik to `[("C", 2), ("D", 1), ("C", 1)]`. Pierwsze dwa C tworzą jedną serię, litera D ją przerywa, a ostatnie C rozpoczyna nową. Nie chodzi o łączną liczbę wszystkich liter C w sygnale.


```python
def spakuj(sygnal):
    if not sygnal:
        return []

    wynik = []
    poprzednia = sygnal[0]
    licznik = 1

    for znak in sygnal[1:]:
        if znak == poprzednia:
            licznik += 1
        else:
            wynik.append((poprzednia, licznik))
            poprzednia = znak
            licznik = 1

    wynik.append((poprzednia, licznik))
    return wynik

print(spakuj("CCDC"))
print(spakuj("AAABBCA"))
print(spakuj(""))
```

## Zadanie 13: anagramy

**Dane:** Dwa napisy `a` i `b`. Nie pomijamy spacji ani nie zmieniamy wielkości liter.

**Do wykonania:** Napisz funkcję `anagramy(a, b)`, która rozstrzyga, czy napisy składają się z dokładnie tych samych znaków użytych tyle samo razy, choć niekoniecznie w tej samej kolejności.

**Wynik:** Zwróć `True`, jeśli można przestawić znaki pierwszego napisu, aby otrzymać drugi, i `False` w przeciwnym przypadku. Pokaż wynik dla par `("kot", "tok")` i `("aa", "ab")`.

**Przykład wyjaśniający:** `"aab"` i `"aba"` są anagramami: oba mają dwa a i jedno b. Natomiast `"aab"` i `"abb"` nie są, chociaż używają tych samych dwóch liter. Wyjaśnij na tym przykładzie, dlaczego sam zbiór znaków nie wystarcza.


```python
def anagramy(a, b):
    if len(a) != len(b):
        return False

    licznik = {}

    for znak in a:
        licznik[znak] = licznik.get(znak, 0) + 1

    for znak in b:
        if znak not in licznik:
            return False
        licznik[znak] -= 1
        if licznik[znak] < 0:
            return False

    return all(licznik[znak] == 0 for znak in licznik)

print(anagramy("aab", "aba"))
print(anagramy("kot", "tok"))
print(anagramy("aa", "ab"))
```

**Do wykonania:** Napisz funkcję `najdluzszy(napisy)`, która wybiera najdłuższy napis. Jeśli kilka ma tę samą największą długość, wybierz pierwszy z nich. Nie sortuj całej listy.

**Wynik:** Zwróć wybrany napis w całości, a nie jego długość ani pozycję. Jeśli lista jest pusta, zwróć `None`. Pokaż działanie dla `["ab", "cd", "x"]` i pustej listy.

**Przykład wyjaśniający:** Dla `["lis", "ryba", "sowa"]` wybieramy `"ryba"`. Zarówno `"ryba"`, jak i `"sowa"` mają cztery znaki, ale `"ryba"` pojawiła się wcześniej. Kolejność alfabetyczna nie rozstrzyga tego remisu.


```python
def najdluzszy(napisy):
    if not napisy:
        return None

    najdluzszy_napis = napisy[0]

    for napis in napisy[1:]:
        if len(napis) > len(najdluzszy_napis):
            najdluzszy_napis = napis

    return najdluzszy_napis

print(najdluzszy(["lis", "ryba", "sowa"]))
print(najdluzszy(["ab", "cd", "x"]))
print(najdluzszy([]))
```
